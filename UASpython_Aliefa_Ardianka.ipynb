```python
print("Selamat Datang di Data Sensor Suhu Mekatronika")

data_suhu = []

while True:
    print("\nMenu Data Sensor Suhu Pada Sistem Mekatronika")
    print(" 1. Tambah Data Sensor Suhu")
    print(" 2. Tampilan Keseluruhan Data Suhu")
    print(" 3. Cari Data Sensor Berdasarkan ID")
    print(" 4. Edit Data Sensor Berdasarkan ID")
    print(" 5. Hapus Data Sensor Berdasarkan ID")
    print(" 6. Analisis Data Sensor Suhu")
    print(" 7. Aku mau keluar dari program ini")
    pilihan = input("Masukkan pilihan Anda: ")

    if pilihan == "1":
        print("\n--- Tambah Data Sensor Suhu ---")
        sensor_id = input("ID Sensor: ")
        nilai_suhu= int(input("Nilai Suhu : "))
        satuan_suhu = "°C"
        status_suhu = "Normal"
        if nilai_suhu <= 30:
            status_suhu = "Masih Adem"
        else:
            status_suhu = "Panas Banget"
        data_suhu.append([sensor_id, nilai_suhu, satuan_suhu, status_suhu])


    if pilihan == "2":
        print("\n--- Tampilan Keseluruhan Data Suhu ---")
        if data_suhu:
            for n, p in enumerate(data_suhu, start=1):
                print(f"Nomor {n} : ID {p[0]} : Nilai {p[1]} {p[2]} : Status {p[3]}")
        else:
            print("Tidak ada data sensor yang terdaftar.")


    elif pilihan == "3":
        print("\n--- Cari Data Sensor Berdasarkan ID ---")
        cari_id = input("Masukkan ID Sensor: ")
        for p in data_suhu:
            if p[0] == cari_id:
                print(f"ID: {p[0]}, Nilai: {p[1]} {p[2]}, Status: {p[3]}")
                break
        else:
          print("Data tidak ada.")


    elif pilihan == "4":
        print("\n--- Edit Data Sensor Berdasarkan ID ---")
        edit_id = input("Masukkan ID Sensor yang akan diedit: ")
        for i, p in enumerate(data_suhu):
            if p[0] == edit_id:
                nilai = int(input("Nilai Suhu baru: "))
                satuan_suhu = "°C"
                status_suhu = "Normal"
                if nilai_suhu <= 30:
                  status_suhu = "Masih Adem"
                else:
                  status_suhu = "Panas Banget"
                data_suhu[i] = [edit_id, nilai, satuan_suhu, status_suhu]
                print("Data berhasil diubah.")
                break
        else:
          print("Data tidak ada.")


    elif pilihan == "5":
        print("\n--- Hapus Data Sensor Berdasarkan ID ---")
        hapus_id = input("Masukkan ID Sensor yang akan dihapus: ")
        for i, p in enumerate(data_suhu):
            if p[0] == hapus_id:
                del data_suhu[i]
                print("Data berhasil dihapus.")
                break
        else:
          print("Data tidak ada.")


    elif pilihan == "6":
        print("\n--- Analisis Data Sensor ---")
        if len(data_suhu) == 0:
            print("Belum ada data untuk dianalisis.")
        else:
            nilai_seluruh = [p[1] for p in data_suhu]
            rata_rata = sum(nilai_seluruh) / len(nilai_seluruh)
            maksimum = max(nilai_seluruh)
            minimum = min(nilai_seluruh)
            print(f"Rata-rata Suhu: {rata_rata:.2f} °C")
            print(f"Suhu Tertinggi: {maksimum} °C")
            print(f"Suhu Terendah: {minimum} °C")


    elif pilihan == "7":
        print("\nProgram telah selesai😁👍. Arigatouunee!")
        break

    else:
        print("Pilihan menu kamu tidak ada.")

```
